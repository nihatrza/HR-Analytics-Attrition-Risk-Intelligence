```python
df.drop(columns=["EmployeeCount", "Over18", "StandardHours"], inplace=True)
```

```python
df.dtypes
```

```python
age_bins = [0, 24, 34, 44, 54, 100]
age_labels = ["Under 25", "25-34", "35-44", "45-54", "55+"]
df["AgeGroup"] = pd.cut(df["Age"], bins=age_bins, labels=age_labels)
```

```python
tenure_bins = [-1, 2, 5, 10, 100]
tenure_labels = [
    "New (0-2 yrs)",
    "Early (3-5 yrs)",
    "Established (6-10 yrs)",
    "Veteran (10+ yrs)",
]
df["TenureGroup"] = pd.cut(
    df["YearsAtCompany"], bins=tenure_bins, labels=tenure_labels
)
```

```python
df["OverallSatisfactionIndex"] = ((
    df["EnvironmentSatisfaction"]
    + df["JobSatisfaction"]
    + df["RelationshipSatisfaction"]
    + df["WorkLifeBalance"]
) / 4).round(2)
```

```python
df["OverallSatisfactionIndex"]
```

```python
df["Attrition_Numeric"] = df["Attrition"].apply(
    lambda x: 1 if x == "Yes" else 0
)
```

```python
df.head()
```

```python
df.columns
```

```python
dim_job = (
    df[["Department", "JobRole", "JobLevel"]]
    .drop_duplicates()
    .reset_index(drop=True)
)


dim_job["job_id"] = dim_job.index + 1


dim_job = dim_job[["job_id", "Department", "JobRole", "JobLevel"]]

dim_job.head()
```

```python
dim_demographics = (
    df[["Gender", "MaritalStatus", "Education", "EducationField", "AgeGroup"]].drop_duplicates().reset_index(drop=True))

dim_demographics["demographics_id"] = dim_demographics.index + 1

dim_demographics = dim_demographics[["demographics_id","Gender","MaritalStatus","Education","EducationField","AgeGroup",]]


dim_demographics.head()
```

```python
dim_satisfaction = (
    df[
        [
            "EnvironmentSatisfaction",
            "JobSatisfaction",
            "RelationshipSatisfaction",
            "WorkLifeBalance",
            "JobInvolvement",
            "PerformanceRating",
            "OverallSatisfactionIndex",
        ]
    ]
    .drop_duplicates()
    .reset_index(drop=True)
)

dim_satisfaction["satisfaction_id"] = dim_satisfaction.index + 1

dim_satisfaction = dim_satisfaction[
    [
        "satisfaction_id",
        "EnvironmentSatisfaction",
        "JobSatisfaction",
        "RelationshipSatisfaction",
        "WorkLifeBalance",
        "JobInvolvement",
        "PerformanceRating",
        "OverallSatisfactionIndex",
    ]
]

dim_satisfaction.head()
```

```python
fact_df = df.merge(
    dim_job, on=["Department", "JobRole", "JobLevel"], how="left"
)
```

```python
fact_df = fact_df.merge(
    dim_demographics,
    on=["Gender", "MaritalStatus", "Education", "EducationField", "AgeGroup"],
    how="left",
)
```

```python
fact_df = fact_df.merge(
    dim_satisfaction,
    on=[
        "EnvironmentSatisfaction",
        "JobSatisfaction",
        "RelationshipSatisfaction",
        "WorkLifeBalance",
        "JobInvolvement",
        "PerformanceRating",
        "OverallSatisfactionIndex",
    ],
    how="left",
)
```

```python
fact_employee_attrition = fact_df[
    [
        "EmployeeNumber", 
        "job_id", 
        "demographics_id",  
        "satisfaction_id",  
        "Age",
        "DailyRate",
        "DistanceFromHome",
        "HourlyRate",
        "MonthlyIncome",
        "MonthlyRate",
        "NumCompaniesWorked",
        "PercentSalaryHike",
        "StockOptionLevel",
        "TotalWorkingYears",
        "TrainingTimesLastYear",
        "YearsAtCompany",
        "YearsInCurrentRole",
        "YearsSinceLastPromotion",
        "YearsWithCurrManager",
        "Attrition",
        "Attrition_Numeric",
        "BusinessTravel",
        "OverTime",
        "TenureGroup",
    ]
]


fact_employee_attrition.head()
```

```python
from sqlalchemy import create_engine

DB_USER = "postgres"
DB_PASSWORD = "admin123" 
DB_HOST = "localhost"
DB_PORT = "5432"
DB_NAME = "hr_analytics" 


connection_str = (
    f"postgresql://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}"
)
engine = create_engine(connection_str)


dim_job.to_sql("dim_job", engine, if_exists="replace", index=False)
dim_demographics.to_sql(
    "dim_demographics", engine, if_exists="replace", index=False
)
dim_satisfaction.to_sql(
    "dim_satisfaction", engine, if_exists="replace", index=False
)
fact_employee_attrition.to_sql(
    "fact_employee_attrition", engine, if_exists="replace", index=False
)
```
